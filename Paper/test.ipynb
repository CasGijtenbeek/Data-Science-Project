```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor
import xgboost

stores = pd.read_csv("stores.csv")
features = pd.read_csv("features.csv")
train = pd.read_csv("train.csv")
```

```python
df = train.merge(features, how='left', on=['Store', 'Date', 'IsHoliday']).merge(stores, how='left', on='Store')
df['IsHoliday'] = df['IsHoliday'].astype(int)
df.drop(['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5'], axis=1, inplace=True)

print(df.head())
```

```python
X_data, Y_data = np.array(df.drop(["Weekly_Sales", "Date", "Type"], axis=1).values), np.array(df["Weekly_Sales"].values)

scaler_X = MinMaxScaler()
xscale = scaler_X.fit_transform(X_data)

print(Y_data)

Y_data = np.reshape(Y_data, (-1, 1))

print(Y_data)
scaler_y = MinMaxScaler()
yscale = scaler_y.fit_transform(Y_data)
```

```python
x_train, x_test, y_train, y_test = train_test_split(xscale, yscale, test_size=0.33, random_state=42)
```

```python
rf = RandomForestRegressor(n_estimators=175, max_depth=28)
rf.fit(x_train, y_train)
y_pred_rf = rf.predict(x_test)
print("Random Forest Model Performance:")
print("R^2 Score:", r2_score(y_test, y_pred_rf))
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred_rf))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred_rf))
```

```python
xgb = xgboost.XGBRegressor(max_depth=28, n_estimators=150)
xgb.fit(x_train, y_train)
y_pred_xgb = xgb.predict(x_test)

print("XGBoost Model Performance:")
print("R^2 Score:", r2_score(y_test, y_pred_xgb))
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred_xgb))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred_xgb))
```
