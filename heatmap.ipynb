```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from ipywidgets import interact, Dropdown

df = pd.read_excel("caffe_change.xlsx")

```

```python



weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
df["Weekday"] = pd.Categorical(df["Weekday"], categories=weekday_order, ordered=True)

categories = sorted(df["Category"].unique().tolist())
categories = ["all"] + categories   # add 'all' option

def plot_heatmap(category):
    if category != "all":
        df_filtered = df[df["Category"] == category]
    else:
        df_filtered = df

    heatmap_df = (
        df_filtered.groupby(["Time Bucket", "Weekday"], observed=True)["Quantity"]
                   .sum()
                   .reset_index(name="Total Quantity")
    )

    pivot_df = heatmap_df.pivot(index="Time Bucket", columns="Weekday", values="Total Quantity").fillna(0)

    plt.figure(figsize=(12, 6))
    sns.heatmap(
        pivot_df,
        annot=True,
        fmt=".0f",
        linewidths=.5,
        cmap="Reds"    
    )

    plt.title(f"Quantity Sold Heatmap (Category = {category})")
    plt.xlabel("Weekday")
    plt.ylabel("Time Bucket")
    plt.tight_layout()
    plt.show()

interact(plot_heatmap, category=Dropdown(options=categories, value="all", description="Category:"));
```
